```python
from google.colab import files
from zipfile import ZipFile
from pathlib import Path
import os, tempfile, subprocess, sys
uploaded = files.upload()
work = Path(tempfile.mkdtemp())
with ZipFile(next(iter(uploaded))) as archive:
    for name in ['investigator.py', 'requirements.txt']:
        (work / name).write_bytes(archive.read('pipeline-investigator/' + name))
os.chdir(work)
```

```python
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'requirements.txt'], check=True)
```

```python
subprocess.run([sys.executable, 'investigator.py', 'train'], check=True)
```

```python
subprocess.run([sys.executable, 'investigator.py', 'demo'], check=True)
```
